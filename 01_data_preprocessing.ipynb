# 📊 01 — Data Preprocessing Pipeline

This notebook demonstrates a complete data preprocessing workflow:
- Handling missing values
- Encoding categorical variables
- Feature scaling
- Train-test split

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.impute import SimpleImputer

print('Libraries loaded ✅')
```

```python
# --- Sample dataset ---
df = pd.DataFrame({
    'age':    [25, np.nan, 35, 45, np.nan, 28],
    'salary': [50000, 60000, np.nan, 80000, 55000, 62000],
    'dept':   ['IT', 'HR', 'IT', np.nan, 'Finance', 'HR'],
    'hired':  [1, 0, 1, 1, 0, 1]
})
print('Shape:', df.shape)
df.head()
```

```python
# --- Impute missing values ---
num_imputer = SimpleImputer(strategy='mean')
df[['age', 'salary']] = num_imputer.fit_transform(df[['age', 'salary']])

cat_imputer = SimpleImputer(strategy='most_frequent')
df[['dept']] = cat_imputer.fit_transform(df[['dept']])

print('Missing values after imputation:')
print(df.isnull().sum())
```

```python
# --- Encode categorical + scale numerics ---
le = LabelEncoder()
df['dept_encoded'] = le.fit_transform(df['dept'])

X = df[['age', 'salary', 'dept_encoded']]
y = df['hired']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
print(f'Train: {X_train.shape}, Test: {X_test.shape}')
```
